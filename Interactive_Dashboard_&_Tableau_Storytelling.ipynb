**Q1 Explain the difference between a Tableau Dashboard and a Tableau Story Point. Give two healthcare-related use cases.**

**Ans-**

**Tableau Dashboard**

*   A Dashboard is a collection of multiple visualizations displayed on a single screen.

*   Used for monitoring KPIs and interactive analysis.
*   Allows users to filter, drill down, and explore data.


*   Focuses on data exploration.

Example: One screen showing patient count, treatment cost, and department performance.

**Tableau Story Point**

*   A Story Point is a sequence of dashboards and worksheets arranged to present insights in a logical narrative.

*  Used for presenting findings and explaining insights step-by-step.
*   Guides users through a specific analytical journey.


*   Focuses on data storytelling and decision-making.

Example: A presentation that starts with patient trends, then department analysis, followed by regional performance and recommendations.



**Q2 What are Actions in Tableau? Explain how Filter Action can help hospital management analyze patient data.**

**Ans-**

Actions are interactive features in Tableau that allow users to interact with dashboards and visualizations. Common actions include Filter, Highlight, URL, and Parameter Actions.

**How Filter Action Helps Hospital Management?**

A Filter Action lets users click on a data point and automatically filter related charts.

**Example:**

 If a manager clicks on the West Region, all dashboard charts update to show:


*   Patients from West Region
*   Treatment Costs

*   Department Performance
*   Patient Outcomes

**Benefit:** Helps hospital management quickly analyze patient trends and make better decisions.

**1. Which department generates the highest treatment cost?**

 Neurology generates the highest treatment cost, totaling approximately ₹5.32 million. This indicates that Neurology requires substantial resources and contributes the most to overall hospital expenditure.

**2. Which region has more critical patients?**

 South Region has the highest number of critical patients (approximately 20 patients), followed by the East and North regions.

**3. What business decisions can hospital management take from these insights?**

**Resource Allocation**

Increase medical staff, equipment, and budget allocation for the Neurology department due to its high treatment costs.

**Critical Care Improvement**

Since the South Region has the highest number of critical patients, management should strengthen ICU facilities, specialist availability, and emergency care services in this region.

**Cost Optimization**

Review treatment procedures and operational expenses in Neurology and Cardiology to identify opportunities for reducing costs without compromising patient care.

 **Performance Management**

Recognize high-performing doctors such as Dr. Mehta and Dr. Iyer, who handled the highest patient volumes, and share their best practices across departments.

Use the cost and outcome insights to improve budgeting, staffing decisions, and healthcare quality across all hospital regions.
